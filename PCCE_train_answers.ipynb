## 1번 · 입력받은 두 수의 합

빈칸: `int`

`input().split()`은 문자열들을 만듭니다. `map`이 각 문자열을 `int`에 전달해서 정수로 바꾸고, 두 값을 `a`, `b`에 저장합니다. 변환 없이 문자열끼리 `+`를 하면 숫자의 합이 아니라 문자열 연결이 됩니다.

예시 답안 검증: 5개 사례 통과.

```python
a, b = map(int, input().split())
print(a + b)
```

## 2번 · 60점도 합격

수정: `if score > 60:` → `if score >= 60:`

`>`는 기준값을 제외하고, `>=`는 기준값도 포함합니다. 60점을 반드시 따로 확인해야 합니다. `print`의 들여쓰기는 각 분기 안에 유지하세요.

예시 답안 검증: 5개 사례 통과.

```python
score = int(input())

if score >= 60:
    print("PASS")
else:
    print("FAIL")
```

## 3번 · n 이하의 양의 짝수

빈칸 1: `n + 1` / 빈칸 2: `2`

`range(시작, 끝, 간격)`은 끝값을 포함하지 않습니다. 따라서 n을 포함할 가능성을 남기려면 끝에 `n + 1`을 넣습니다. 2부터 2씩 증가하면 양의 짝수가 나옵니다.

예시 답안 검증: 5개 사례 통과.

```python
n = int(input())
result = list(range(2, n + 1, 2))
print(result)
```

## 4번 · 입력된 단어 반복하기

수정: `print("word", end="")` → `print(word, end="")`

따옴표로 감싼 `"word"`는 고정된 글자입니다. 따옴표 없는 `word`는 입력값을 저장한 변수입니다. `end=""`는 매번 출력 뒤에 줄바꿈을 붙이지 않도록 합니다. 입력이 달라지면 출력할 단어도 달라져야 합니다.

예시 답안 검증: 5개 사례 통과.

```python
word, n = input().split()
n = int(n)

for i in range(n):
    print(word, end="")
```

## 5번 · while로 1부터 n까지 더하기

빈칸 1: `i <= n` / 빈칸 2: `i += 1`

현재 `i`를 합계에 더한 뒤, `i`를 1 증가시킵니다. n까지 더하려면 n과 같을 때도 반복해야 합니다. `i`를 바꾸지 않으면 조건이 계속 참이어서 반복이 끝나지 않을 수 있습니다.

예시 답안 검증: 5개 사례 통과.

```python
n = int(input())
i = 1
total = 0

while i <= n:
    total += i
    i += 1

print(total)
```

## 6번 · 처음 나온 90점 미만에서 멈추기

수정: `idx -= 1` → `idx += 1`

리스트를 왼쪽부터 읽으려면 인덱스가 0, 1, 2, …로 증가해야 합니다. 이 코드는 `print`를 먼저 실행하므로 90점 미만인 점수도 출력합니다. 그다음 `break`가 반복문을 끝냅니다. `idx -= 1`은 뒤쪽 원소를 읽거나 인덱스 오류를 일으킬 수 있습니다.

예시 답안 검증: 6개 사례 통과.

```python
scores = list(map(int, input().split()))
idx = 0

while idx < len(scores):
    print(scores[idx])
    if scores[idx] < 90:
        break
    idx += 1
```

## 7번 · 짝수만 골라 10 더하기

빈칸 1: `x + 10` / 빈칸 2: `x % 2 == 0`

맨 앞은 리스트에 넣을 값, `if` 뒤는 통과 조건입니다. `x % 2 == 0`으로 짝수를 고르고, 통과한 x에 10을 더해 넣습니다. `if x + 10`이라고 하면 이 문제의 범위에서는 계산 결과가 모두 0이 아니어서 모든 원소가 통과합니다.

예시 답안 검증: 5개 사례 통과.

```python
numbers = list(map(int, input().split()))
result = [x + 10 for x in numbers if x % 2 == 0]
print(result)
```

## 8번 · 같은 위치의 값 더하기

빈칸 1: `x + y` / 빈칸 2: `list(result)`

`map`이 a와 b에서 값을 하나씩 꺼내 각각 x와 y로 전달합니다. 람다는 두 값을 더해서 반환합니다. `list(result)`가 결과를 꺼내 리스트로 만듭니다. `print(map)`은 결과 변수가 아닌 내장 `map` 자체를 출력합니다. 같은 map 객체를 이미 모두 꺼내 썼다면 다시 `list`로 바꿀 때 빈 리스트가 되므로, 비교할 때는 셀 전체를 다시 실행하세요.

예시 답안 검증: 5개 사례 통과.

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

result = map(lambda x, y: x + y, a, b)
print(list(result))
```

## 9번 · 기준 점수 이상인 사람 수

처음에는 합격자 수를 0으로 두고, 각 점수가 기준 이상일 때만 1을 더합니다. **모든 점수를 확인한 뒤** 합격자 수를 반환합니다. 따라서 `return count`는 `for` 바깥이면서 함수 안에 있어야 합니다. 함수 내부에서 출력만 하고 반환하지 않으면, 호출 결과는 `None`이 됩니다. 예제 셀 출력은 차례로 `2`, `0`, `0`입니다.

예시 답안 검증: 6개 사례 통과.

```python
def count_passed(scores, cutoff):
    count = 0
    for score in scores:
        if score >= cutoff:
            count += 1
    return count

# 예제 확인
print(count_passed([59, 60, 80], 60))
print(count_passed([30, 40], 50))
print(count_passed([], 60))
```

## 10번 · 예산 안에서 순서대로 구매하기

현재 가격이 남은 예산보다 크면 `break`로 멈춥니다. 구매 가능하면 예산을 줄이고 구매 개수를 늘립니다. `price >= budget`으로 비교하면 예산과 가격이 같은 상품을 잘못 제외합니다. 살 수 없는 상품을 건너뛰는 것도 이 문제의 규칙과 다릅니다. 예제 셀 출력은 차례로 `2`, `0`, `2`, `0`입니다.

예시 답안 검증: 10개 사례 통과.

```python
def buy_count(prices, budget):
    count = 0
    for price in prices:
        if price > budget:
            break
        budget -= price
        count += 1
    return count

# 예제 확인
print(buy_count([3, 4, 5], 7))
print(buy_count([8, 2, 1], 5))
print(buy_count([2, 3], 10))
print(buy_count([], 10))
```
