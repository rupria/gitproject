```python
# 함수 정의

def add (a, b):
    """두 개의 숫자를 더해서 그 값을 반환합니다."""
    return a+b;

def subtract(a,b):
    """두 개의 숫자를 빼서 그 값을 반환합니다."""
    return a-b;

def multiply(a,b):
    """두 개의 숫자를 곱해서 그 값을 반환합니다."""
    return a*b;

def divide(a,b):
    """두 개의 숫자를 나눠서 그 값을 반환합니다."""
    return a/b;
```

```python
# 함수 호출
# 함수의 이름 add, subtract, multiply, divide

add_result=add(5,3) # add에 5와 3을 넣어요.
print("add_result :",add_result) # 그러면 a+b 의값을 프린트 합니다.

subtract_result=subtract(5,3) # 
print("subtract_result :",subtract_result) # 

multiply_result=multiply(a=5,b=3) # 
print("multiply_result :",multiply_result) # 


divide_result=divide(5, b=3) # 
print("divide_result :",divide_result) # 
```
