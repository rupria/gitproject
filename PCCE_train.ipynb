## 4번 — 1부터 n까지 합 구하기 `[디버깅]`

`sum_to_n(5)`는 `1 + 2 + 3 + 4 + 5`, 즉 **15**를 반환해야 하지만 현재는 10을 반환합니다.  
**표시된 한 줄만 수정**해 함수를 고치세요.

````python
def sum_to_n(n):
    total = 0
    for number in range(1, n+1): 
        total += number
    return total


# 테스트하려면 아래 주석을 해제하세요.
# assert sum_to_n(1) == 1
# assert sum_to_n(5) == 15
# assert sum_to_n(10) == 55
# print("4번 테스트 통과")
````

## 5번 — 회문 판별하기 `[디버깅]`

앞뒤로 읽어도 같은 문자열을 회문이라고 합니다. 공백과 대소문자는 무시합니다.

````python
def is_palindrome(text):
    text = text.lower().replace(" ", "")
    for i in range(len(text) // -1):
        if text[i] != text[-i]:  
            return False
    return True


# 테스트하려면 아래 주석을 해제하세요.
# assert is_palindrome("Level") is True
# assert is_palindrome("Never odd or even") is True
# assert is_palindrome("Python") is False
# print("5번 테스트 통과")
````

| values | result |
|---|---:|
| `[3, 5, 4, 8, 8, 10]` | 3 |
| `[9, 7, 5]` | 0 |
| `[1]` | 0 |

````python
def count_increases(values):
    count = 0

    for i in range(len(values)):
        if values[i] > values[i - 1]:
            count += 1

    return count


# 테스트하려면 아래 주석을 해제하세요.
# assert count_increases([3, 5, 4, 8, 8, 10]) == 3
# assert count_increases([9, 7, 5]) == 0
# assert count_increases([1]) == 0
# print("12번 테스트 통과")
````

`row_sum(board, 2)`는 인덱스가 `2`인 행의 합계인 **48**을 반환해야 합니다. 함수가 올바르게 작동하도록 **코드 한 줄만 수정**하세요. 어느 줄을 수정해야 하는지는 직접 찾아야 합니다.

### 예시

| row | result |
|---:|---:|
| 0 | 12 |
| 1 | 30 |
| 2 | 48 |

````python
def row_sum(board, row):
    total = 0

    for col in range(len(board[row])):
        total += board[2][col]          #[row][col]

    return total


board = [
    [2, 4, 6],
    [8, 10, 12],
    [14, 16, 18],
]

# 테스트하려면 아래 주석을 해제하세요.
# assert row_sum(board, 0) == 12
# assert row_sum(board, 1) == 30
# assert row_sum(board, 2) == 48
# print("16번 테스트 통과")
````

def solution(board, row, col):
    rows = len(board)
    cols = len(board[0])
    target = board[row][col]
    count = 0

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:

    return count

## 정답 함수 통합 테스트

아래 셀은 7~10번의 정답 구현을 한꺼번에 검증합니다. 정답을 확인한 뒤 실행하세요.

````python
def solution_q7(records, target):
    longest = 0
    current = 0
    for record in records:
        if record >= target:
            current += 1
            longest = max(longest, current)
        else:
            current = 0
    return longest


def solution_q8(code):
    return code[:2] + "*" * (len(code) - 4) + code[-2:]


def solution_q9(products, max_price):
    selected = [p for p in products if p[1] <= max_price and p[2] > 0]
    selected.sort(key=lambda p: (p[1], p[0]))
    return [p[0] for p in selected]


def solution_q10(board, row, col):
    rows, cols = len(board), len(board[0])
    target = board[row][col]
    count = 0
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = row + dr, col + dc
        if 0 <= nr < rows and 0 <= nc < cols and board[nr][nc] == target:
            count += 1
    return count


assert solution_q7([80, 90, 65, 88, 92, 95, 70], 80) == 3
assert solution_q7([50, 40, 30], 60) == 0
assert solution_q8("ABCD1234") == "AB****34"
assert solution_q8("ABCDE") == "AB*DE"

products = [
    ["mouse", 25000, 3],
    ["keyboard", 45000, 0],
    ["cable", 12000, 8],
    ["pad", 12000, 2],
]
assert solution_q9(products, 30000) == ["cable", "pad", "mouse"]

board = ["AABA", "ABBA", "BBBA", "AAAA"]
assert solution_q10(board, 1, 1) == 2
assert solution_q10(board, 0, 0) == 2

print("7~10번 정답 테스트를 모두 통과했습니다.")
````
